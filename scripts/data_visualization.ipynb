# Data Visualization

on traing dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.manifold import TSNE
```

```python
# Read the training dataset
data = pd.read_csv('../data/training_data.csv')

data.head()
```

## `Diabetes` (label column)

```python
# Distribution of 'Diabetes' label
plt.figure(figsize=(6,4))
data['Diabetes'].value_counts().plot(kind='bar')
plt.title('Distribution of Diabetes Label')
plt.xlabel('Diabetes')
plt.ylabel('Count')
plt.show()
```

## Binary features

`HighBP`, `HighChol`, `CholCheck`, `Smoker`, `Stroke`, `HeartDiseaseorAttack`, `PhysActivity`, <br>
`HvyAlcoholConsump`, `AnyHealthcare`, `NoDocbcCost`, `DiffWalk`, `Sex`

```python
# Distribution of binary features
bin_features = ['HighBP', 'HighChol', 'CholCheck', 'Smoker', 'Stroke',
                'HeartDiseaseorAttack', 'PhysActivity', 'HvyAlcoholConsump',
                'AnyHealthcare', 'NoDocbcCost', 'DiffWalk', 'Sex']

fig = plt.figure(figsize=(12, 8))
for i, feature in enumerate(bin_features, 1):
    plt.subplot(3, 4, i)
    counts = data[feature].value_counts(sort=False).sort_index()
    counts.plot(kind='bar')
    plt.title(f'Distribution of {feature}')
    plt.xlabel(feature)
    plt.ylabel('Count')
plt.tight_layout()
plt.show()
```
