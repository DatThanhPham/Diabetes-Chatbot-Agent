# Read data files

```python
# Read csv files
df2020 = pd.read_csv('diabetes_BRFSS2020.csv')
df2021 = pd.read_csv('diabetes_BRFSS2021.csv')
df2022 = pd.read_csv('diabetes_BRFSS2022.csv')
df2023 = pd.read_csv('diabetes_BRFSS2023.csv')
df2024 = pd.read_csv('diabetes_BRFSS2024.csv')
```

# Double check for missing or duplicate values

```python
# Check for missing values
for year, df in zip(range(2020, 2025), [df2020, df2021, df2022, df2023, df2024]):
    missing_values = df.isnull().sum()
    if missing_values.any():
        print(f"\nMissing values in {year} DataFrame:")
        print(missing_values[missing_values > 0])
    else:
        print(f"\nNo missing values in {year} DataFrame.")
```

```python
# Check for duplicate rows
for year, df in zip(range(2020, 2025), [df2020, df2021, df2022, df2023, df2024]):
    duplicate_rows = df.duplicated().sum()
    if duplicate_rows > 0:
        print(f"\nNumber of duplicate rows in {year} DataFrame: {duplicate_rows}")
    else:
        print(f"\nNo duplicate rows in {year} DataFrame.")
```

```python
# Remove duplicate rows
df2020 = df2020.drop_duplicates()
df2021 = df2021.drop_duplicates()
df2022 = df2022.drop_duplicates()
df2023 = df2023.drop_duplicates()
df2024 = df2024.drop_duplicates()
```

```python
# Check sizes after removing duplicates
print(f"\nSize of 2020 DataFrame after removing duplicates: {df2020.shape}")
print(f"Size of 2021 DataFrame after removing duplicates: {df2021.shape}")
print(f"Size of 2022 DataFrame after removing duplicates: {df2022.shape}")
print(f"Size of 2023 DataFrame after removing duplicates: {df2023.shape}")
print(f"Size of 2024 DataFrame after removing duplicates: {df2024.shape}")
```

# Remove outliers using IQR

```python
# Remove outliers using IQR
def remove_outliers_iqr(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return df[(df >= lower_bound) & (df <= upper_bound)]

numeric_cols = ['BMI', 'MentHlth', 'PhysHlth']
for year, df in zip(range(2020, 2025), [df2020, df2021, df2022, df2023, df2024]):
    for col in numeric_cols:
        df = remove_outliers_iqr(df, col)
    print(f"\nSize of {year} DataFrame after removing outliers: {df.shape}")
```

# Saving cleaned data files

```python
# Save cleaned data files
df2020.to_csv('cleaned_diabetes_BRFSS2020.csv', index=False)
df2021.to_csv('cleaned_diabetes_BRFSS2021.csv', index=False)
df2022.to_csv('cleaned_diabetes_BRFSS2022.csv', index=False)
df2023.to_csv('cleaned_diabetes_BRFSS2023.csv', index=False)
df2024.to_csv('cleaned_diabetes_BRFSS2024.csv', index=False)
```

```python
# Merge and save 2021 + 2023 cleaned data for training
merged_df = pd.concat([df2021, df2023], ignore_index=True)

# Triple check for duplicates in merged data
duplicate_rows_merged = merged_df.duplicated().sum()
if duplicate_rows_merged > 0:
    print(f"\nNumber of duplicate rows in merged DataFrame: {duplicate_rows_merged}")
    merged_df = merged_df.drop_duplicates()
    print(f"Size of merged DataFrame after removing duplicates: {merged_df.shape}")
else:
    print("\nNo duplicate rows in merged DataFrame.")

# Save merged data
merged_df.to_csv('training_data.csv', index=False)
```
