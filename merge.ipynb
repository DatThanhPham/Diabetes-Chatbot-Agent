```python
import pandas as pd
import os

# Danh sách file
data_dir = "data"

files = [
    "cleaned_diabetes_BRFSS2020.csv",
    "cleaned_diabetes_BRFSS2021.csv",
    "cleaned_diabetes_BRFSS2022.csv",
    "cleaned_diabetes_BRFSS2023.csv",
    "cleaned_diabetes_BRFSS2024.csv"
]

file_paths = [os.path.join(data_dir, f) for f in files]

# Đọc tất cả file
dfs = [pd.read_csv(fp) for fp in file_paths]

# Tìm các cột chung
common_cols = set(dfs[0].columns)
for df in dfs[1:]:
    common_cols = common_cols.intersection(df.columns)

common_cols = list(common_cols)
common_cols.sort()  # Cho đẹp và ổn định thứ tự

print("Các cột chung giữa tất cả file:")
print(common_cols)

# Merge các file chỉ theo cột chung
merged_df = pd.concat([df[common_cols] for df in dfs], ignore_index=True)

# Xuất file
output_path = os.path.join(data_dir, "merged_diabetes_dataset.csv")
merged_df.to_csv(output_path, index=False)

print("Merge hoàn tất, file lưu tại:", output_path)
print("Tổng số dòng:", len(merged_df))
```
